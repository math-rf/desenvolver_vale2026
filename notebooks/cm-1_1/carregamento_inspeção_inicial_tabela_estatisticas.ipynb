```python
import pandas as pd

# Define the path to the parquet file
parquet_file_path = '/content/drive/MyDrive/hackathon-vale/pre-processado/telemetria/final_telemetry.parquet'

# Load the parquet file into a pandas DataFrame
df = pd.read_parquet(parquet_file_path)

print(f"Successfully loaded data from {parquet_file_path}. Shape: {df.shape}")
```

```python
summary_stats = []

for column in df.columns:

    if pd.api.types.is_numeric_dtype(df[column]) and not column.startswith('Id'):

        col_min = df[column].min()
        col_max = df[column].max()
        col_mean = df[column].mean()
        col_median = df[column].median()
        col_std = df[column].std()
        col_type = df[column].dtype
        null_percentage = df[column].isnull().sum() / len(df) * 100

        summary_stats.append({
            'Feature': column,
            'DataType': col_type,
            'NullPercentage': f'{null_percentage:.2f}%',
            'Min': col_min,
            'Max': col_max,
            'Mean': col_mean,
            'Median': col_median,
            'StdDev': col_std
        })

stats_df = pd.DataFrame(summary_stats)

display(stats_df)
```
