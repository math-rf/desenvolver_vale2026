# Import dos dados e bibliotecas

````python
import pandas as pd
import numpy as np

import seaborn as sns
import matplotlib.pyplot as plt
import matplotlib.lines as mlines
import pyarrow
````

````python
df_caminhao = pd.read_parquet('datasets/datasets_tratados/df_caminhao_pre_processamento.parquet', engine="pyarrow")
# df_escavadeira = pd.read_parquet('datasets/datasets_tratados/df_escavadeira_pre_processamento.parquet', engine="pyarrow")
````

# Avaliando erros e remoção de ruído

````python
filtro_classe = ~(df_caminhao['Classe'] == 'inactivate')
filtro_nome = ~(df_caminhao['Alarme'].str.contains('inactive'))

df_caminhao = df_caminhao[(filtro_classe) & (filtro_nome)]
````

````python
df_caminhao
````

# Definindo target e refinamento

````python
# 1. Garantir ordenação estrita por equipamento e data
df_caminhao = df_caminhao.sort_values(by=['TAG', 'Data_Evento']).reset_index(
    drop=True
)

# Criamos um array para armazenar os tempos em minutos
minutos_ate_falha = np.full(len(df_caminhao), np.nan)

# 2. Processar por TAG para isolar os equipamentos
for tag, group in df_caminhao.groupby('TAG'):
    indices_grupo = group.index

    # Datas onde Is_Dont_Go == 1
    datas_dont_go = group[group['Is_Dont_Go'] == 1]['Data_Evento'].values

    if len(datas_dont_go) == 0:
        continue  # Máquinas sem falha continuam como NaN

    # Forçamos a busca a olhar estritamente para o futuro (+1 segundo)
    # para que o próprio Don't Go não olhe para si mesmo
    momentos_busca = group['Data_Evento'] + pd.Timedelta(seconds=1)

    # Encontra o índice do próximo Don't Go no futuro
    indices_busca = np.searchsorted(datas_dont_go, momentos_busca.values)

    # Identifica quem não tem mais nenhuma falha dali para a frente
    sem_futuro = indices_busca >= len(datas_dont_go)
    indices_busca[sem_futuro] = len(datas_dont_go) - 1

    # Resgata a data do próximo Don't Go
    proximas_datas = datas_dont_go[indices_busca]

    # Calcula a diferença de tempo absoluta
    diferenca_tempo = proximas_datas - group['Data_Evento'].values

    # Converte a diferença de tempo para minutos (float)
    minutos = diferenca_tempo / np.timedelta64(1, 'm')

    # Quem não tem falha no futuro recebe NaN (ou você pode dropar depois)
    minutos[sem_futuro] = np.nan

    # Se a própria linha for um Don't Go, também desconsideramos (vira NaN)
    # porque queremos o tempo dos erros comuns até a quebra
    # minutos[group['Is_Dont_Go'] == 1] = np.nan

    # Salva no array global
    minutos_ate_falha[indices_grupo] = minutos

# Adiciona a nova coluna na base
df_caminhao['Minutos_Ate_Proximo_Dont_Go'] = minutos_ate_falha

df_caminhao.dropna(subset=['Minutos_Ate_Proximo_Dont_Go'], inplace=True)

df_caminhao
````

````python
# Definimos o limite da nossa "zona de risco/crise" (ex: 12 horas = 720 minutos)
limite_minutos_risco = 720 

# =====================================================================
# ANÁLISE COMPLETA DE ALARMES INFORMACIONAIS E SEM PODER EXPLICATIVO
# =====================================================================

# 1. Agrupamos por tipo de alarme para calcular as métricas de criticidade e crise
analise_alarmes = df_caminhao.groupby('Alarme').agg(
    frequencia_total=('Alarme', 'count'),
    # Conta quantas vezes a criticidade cadastrada foi 'informacional'
    qtd_informacional=('Criticidade', lambda x: (x == 'Informacional').sum()),
    # Quantas vezes esse alarme aconteceu na zona de crise (<= 12 horas antes do Don't Go)
    ocorrencias_na_crise=('Minutos_Ate_Proximo_Dont_Go', lambda x: (x <= limite_minutos_risco).sum()),
    # Mediana do tempo até a falha
    mediana_minutos=('Minutos_Ate_Proximo_Dont_Go', 'median')
).reset_index()

# 2. Calculamos as taxas percentuais reais
analise_alarmes['percentual_criticidade_informacional'] = (analise_alarmes['qtd_informacional'] / analise_alarmes['frequencia_total']) * 100
analise_alarmes['taxa_presenca_crise_percentual'] = (analise_alarmes['ocorrencias_na_crise'] / analise_alarmes['frequencia_total']) * 100

# 3. FILTRO CRÍTICO: 
# Queremos alarmes que são majoritariamente informacionais (ex: > 90%) 
# E que possuem TAXA DE PRESENÇA NA CRISE IGUAL A 0% (nunca ocorrem perto da falha)
# Exigimos também uma frequência mínima de 10 ocorrências para validação estatística
filtro_ruido = (
    (analise_alarmes['percentual_criticidade_informacional'] == 100.0) & (analise_alarmes['taxa_presenca_crise_percentual'] <= 1.0) | 
    (analise_alarmes['percentual_criticidade_informacional'] == 100.0) & (analise_alarmes['frequencia_total'] >= 75000) | 
    (analise_alarmes['taxa_presenca_crise_percentual'] <= 1.0) & (analise_alarmes['frequencia_total'] >= 100)
)

alarmes_ruido_descarte = analise_alarmes[filtro_ruido].sort_values(by='frequencia_total', ascending=False)

# 4. Exibindo os resultados para validação
print(f"🎯 Sucesso! Encontrados {len(alarmes_ruido_descarte)} tipos de alarmes que são ruídos puros.")

alarmes_ruido_descarte
````

Com base na lista de alarmes identificados pelos critérios de frequência e proporção de ocorrência próximo de crise buscamos a lista de erros e procuramos erros que potencialmente são perigosos e portanto não serão removidos os erros:

- ```6th tire pressure very low```
- ```5th tire pressure high maximum```
- ```final drive bypass solenoid - active```

````python
alarmes_remover = list(alarmes_ruido_descarte['Alarme'].unique())

alarmes_remover.remove("6th tire pressure very low")
alarmes_remover.remove("5th tire pressure high maximum")
alarmes_remover.remove("final drive bypass solenoid - active")

index_remover = df_caminhao[df_caminhao['Alarme'].isin(alarmes_remover)].index

df_caminhao = df_caminhao.drop(index_remover).reset_index(drop=True)
````

````python
def plot_serie_temporal_tag(tag_selecionada):

    df_tag = df_caminhao[df_caminhao['TAG'] == tag_selecionada].copy()

    # Garantimos que está ordenado por data
    df_tag = df_tag.sort_values(by='Data_Evento')

    # Contagem cumulativa contínua global
    df_tag['contagem_cumulativa'] = range(1, len(df_tag) + 1)

    # ==========================================
    # MAPEAMENTO DE CORES PARA OS STATUS
    # ==========================================
    # Ajuste os nomes das chaves de acordo com as strings exatas da sua coluna
    cores_status = {
        'Operando': '#2ecc71',  # Verde
        'Manutenção': '#3498db',  # Azul
        'Parado': '#95a5a6',
        'Hibernando': '#d3d3d3',  # Cinza
    }

    # ==========================================
    # 3. PLOTANDO O GRÁFICO SEPARADO POR MÊS
    # ==========================================
    sns.set_theme(style="whitegrid")

    # Criamos a grade separando pela coluna 'Mes'
    g = sns.FacetGrid(
        df_tag,
        col="Mes",
        col_wrap=3,
        sharex=False,
        sharey=False,  # Mantém o foco na faixa acumulada de cada mês
        height=4,
        aspect=1.3,
    )

    # Passo A: Linha mestre em degraus (esqueleto neutro do acúmulo)
    g.map_dataframe(
        sns.lineplot,
        x="Data_Evento",
        y="contagem_cumulativa",
        drawstyle='steps-pre',
        color='#bdc3c7',
        alpha=0.6,
        linewidth=1.5,
    )

    # Passo B: Função customizada para plotar as situações e os Don't Go
    def plot_detalhes_operacionais(data, **kwargs):
        if data.empty:
            return

        # 1. Plota todos os erros como pontinhos coloridos pelo status operacional atual
        # Usamos um tamanho controlado (s=25) para não sobrepor totalmente a linha
        # IMPORTANTE: Altere 'Situacao_Operacional' para o nome real da sua coluna de status
        sns.scatterplot(
            data=data,
            x="Data_Evento",
            y="contagem_cumulativa",
            hue="Situacao_Operacional",
            palette=cores_status,
            s=15,
            alpha=0.8,
            edgecolor=None,
            zorder=4,
        )

        # 2. Aplica o 'X' Vermelho Soberano exatamente onde houve o Don't Go
        df_critico = data[data['Is_Dont_Go'] == 1]
        if not df_critico.empty:
            plt.scatter(
                df_critico['Data_Evento'],
                df_critico['contagem_cumulativa'],
                color='red',
                marker='X',  # O 'X' destaca muito mais que uma bolinha
                s=50,  # Bem maior para sobressair no meio dos outros pontos
                zorder=5,
            )

    # Aplicando a nossa função lógica na grade
    g.map_dataframe(plot_detalhes_operacionais)

    # Ajustes de estética dos eixos
    g.set_axis_labels("Data do Evento", "Total de Alarmes Acumulados")
    g.set_titles(col_template="Mês: {col_name}")

    # Rotacionando as datas no eixo X para ficar legível
    for ax in g.axes.flat:
        for label in ax.get_xticklabels():
            label.set_rotation(45)

    # ==========================================
    # CRIAÇÃO DA LEGENDA MANUAL GLOBAL
    # ==========================================
    # Construímos a legenda do zero para evitar duplicações e bugs de mapeamento do FacetGrid
    legend_elements = [
        mlines.Line2D(
            [0],
            [0],
            color='#bdc3c7',
            linewidth=2,
            label='Tendência Global de Alarmes',
        ),
        mlines.Line2D(
            [0],
            [0],
            marker='o',
            color='w',
            markerfacecolor='#2ecc71',
            markersize=8,
            label='Status: Operação',
        ),
        mlines.Line2D(
            [0],
            [0],
            marker='o',
            color='w',
            markerfacecolor='#3498db',
            markersize=8,
            label='Status: Manutenção',
        ),mlines.Line2D(
            [0],
            [0],
            marker='o',
            color='w',
            markerfacecolor='#95a5a6',
            markersize=8,
            label='Status: Parado',
        ),
        mlines.Line2D(
            [0],
            [0],
            marker='o',
            color='w',
            markerfacecolor='#d3d3d3',
            markersize=8,
            label='Status: Hibernando',
        ),
        mlines.Line2D(
            [0],
            [0],
            marker='X',
            color='w',
            markerfacecolor='red',
            markeredgecolor='red',
            markersize=9,
            label="Evento 'Don't Go'",
        ),
    ]

    # Joga a legenda para fora dos gráficos, centralizada à direita
    g.fig.legend(
        handles=legend_elements,
        loc='center left',
        bbox_to_anchor=(1.01, 0.5),
        title="Dicionário de Eventos",
    )

    plt.suptitle(
        f"Análise de Alarmes por Estado Operacional - Equipamento: {tag_selecionada}",
        y=1.05,
        fontsize=16,
    )

    # Ajusta o layout deixando uma margem na direita (rect) para a legenda não ser cortada
    plt.tight_layout(rect=[0, 0, 0.9, 1])
    plt.show()
````

````python
tags = df_caminhao['TAG'].unique()

for tag in tags:
    plot_serie_temporal_tag(tag)
````

````python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Configurando o estilo visual do Seaborn
sns.set_theme(style="whitegrid")

# Criando a área para 2 gráficos lado a lado
fig, axes = plt.subplots(1, 2, figsize=(18, 6))

# =====================================================================
# GRÁFICO 1: Top 15 Operadores com Mais 'Don't Go' (Absoluto)
# =====================================================================
df_dont_go = df_caminhao[df_caminhao['Is_Dont_Go'] == 1]

# Agrupamos e pegamos os 15 que mais aparecem
top_operadores_abs = df_dont_go["Nome_Operador_Anon"].value_counts().head(15).reset_index()
top_operadores_abs.columns = ["Nome_Operador_Anon", 'Quantidade_Dont_Go']

sns.barplot(
    data=top_operadores_abs,
    y="Nome_Operador_Anon",
    x='Quantidade_Dont_Go',
    palette="Reds_r", # Cores degradê vermelhas para destacar o topo
    ax=axes[0],
    edgecolor="black"
)

axes[0].set_title("Top 15 Operadores com Maior Volume Absoluto de 'Don't Go'", fontsize=13, fontweight='bold')
axes[0].set_xlabel("Quantidade Total de Eventos 'Don't Go'")
axes[0].set_ylabel("Identificação do Operador")

# Adiciona os valores no fim de cada barra
for p in axes[0].patches:
    width = p.get_width()
    axes[0].annotate(
        f'{int(width):,}',
        (width, p.get_y() + p.get_height() / 2.),
        ha='left', 
        va='center', 
        xytext=(5, 0), 
        textcoords='offset points',
        fontsize=10
    )

# =====================================================================
# GRÁFICO 2: Top 15 Operadores por Taxa de 'Don't Go' (Proporcional)
# =====================================================================
# Para evitar operadores que trabalharam apenas 5 minutos e tiveram 1 erro (taxa de 100%),
# filtramos apenas operadores que geraram um volume mínimo de logs gerais (ex: pelo menos 500 erros)
minimo_logs = 500
contagem_geral = df_caminhao["Nome_Operador_Anon"].value_counts()
operadores_frequentes = contagem_geral[contagem_geral >= minimo_logs].index

df_filtrado_op = df_caminhao[df_caminhao["Nome_Operador_Anon"].isin(operadores_frequentes)]

# Calculamos a taxa: (soma de dont_go / total de logs) * 100
taxa_operador = (
    df_filtrado_op.groupby("Nome_Operador_Anon")['Is_Dont_Go']
    .mean() * 100
).reset_index()

taxa_operador.columns = ["Nome_Operador_Anon", 'Taxa_Dont_Go_Percentual']
top_operadores_taxa = taxa_operador.sort_values(by='Taxa_Dont_Go_Percentual', ascending=False).head(15)

sns.barplot(
    data=top_operadores_taxa,
    y="Nome_Operador_Anon",
    x='Taxa_Dont_Go_Percentual',
    palette="Oranges_r",
    ax=axes[1],
    edgecolor="black"
)

axes[1].set_title(f"Top 15 por Taxa de 'Don't Go' (mínimo {minimo_logs} logs gerados)", fontsize=13, fontweight='bold')
axes[1].set_xlabel("Taxa de Erros que viram 'Don't Go' (%)")
axes[1].set_ylabel("") # Remove para não repetir o rótulo do eixo Y

# Adiciona as porcentagens no fim de cada barra
for p in axes[1].patches:
    width = p.get_width()
    axes[1].annotate(
        f'{width:.2f}%',
        (width, p.get_y() + p.get_height() / 2.),
        ha='left', 
        va='center', 
        xytext=(5, 0), 
        textcoords='offset points',
        fontsize=10
    )

# Título Geral
plt.suptitle("Análise de Influência do Fator Humano nos Eventos de Falha", y=1.02, fontsize=16, fontweight='bold')
plt.tight_layout()
plt.show()
````

````python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Configurando o estilo visual do Seaborn
sns.set_theme(style="whitegrid")

# Criando a área do gráfico com 2 subplots lado a lado
fig, axes = plt.subplots(1, 2, figsize=(16, 6))

# Filtramos os valores nulos para não dar erro no plot
dados_plot = df_caminhao['Minutos_Ate_Proximo_Dont_Go'].dropna()

# Convertendo para horas para facilitar a interpretação operacional
horas_ate_falha = dados_plot / 60

# =====================================================================
# GRAFICO 1: Distribuição Total (Visão Geral)
# =====================================================================
sns.histplot(
    horas_ate_falha,
    bins=50,
    kde=True,
    color='#34495e',
    ax=axes[0]
)
axes[0].set_title("Distribuição Completa (em Horas)", fontsize=13, fontweight='bold')
axes[0].set_xlabel("Tempo até o próximo Don't Go (Horas)")
axes[0].set_ylabel("Frequência (Quantidade de Logs)")

# =====================================================================
# GRAFICO 2: Zoom nas primeiras 48 horas (Foco na Degradação Ativa)
# =====================================================================
# Filtramos apenas eventos que acontecem em até 48 horas (2880 minutos)
horas_limite = 48
dados_zoom = horas_ate_falha[horas_ate_falha <= horas_limite]

sns.histplot(
    dados_zoom,
    bins=48, # 1 bin por hora para ficar bem detalhado
    kde=True,
    color='#e74c3c', # Vermelho para destacar a zona crítica
    ax=axes[1]
)
# Adiciona linhas de referência para ajudar na tomada de decisão operacional
axes[1].axvline(x=6, color='black', linestyle='--', alpha=0.7, label='Janela de 6h')
axes[1].axvline(x=12, color='blue', linestyle='--', alpha=0.7, label='Janela de 12h')

axes[1].set_title(f"Foco nas Primeiras {horas_limite} Horas (Zona de Risco)", fontsize=13, fontweight='bold')
axes[1].set_xlabel("Tempo até o próximo Don't Go (Horas)")
axes[1].set_ylabel("Frequência (Quantidade de Logs)")
axes[1].legend()

# Título Geral do Painel
plt.suptitle("Análise de Dispersão: Tempo Restante até o Próximo Evento 'Don't Go'", y=1.02, fontsize=16, fontweight='bold')

plt.tight_layout()
plt.show()
````
